### Objective:
Train and benchmark multiple classification pipelines with TF-IDF n-grams to determine the most effective classifier.


```python
import os
import pandas as pd
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression, SGDClassifier
from sklearn.naive_bayes import MultinomialNB
from sklearn.svm import LinearSVC
from sklearn.calibration import CalibratedClassifierCV
from sklearn.metrics import classification_report, confusion_matrix, f1_score
```

```python
train_df = pd.read_csv(os.path.join('..', 'data', 'processed', 'train.csv'))
test_df = pd.read_csv(os.path.join('..', 'data', 'processed', 'test.csv'))

X_train, y_train = train_df['text'], train_df['label']
X_test, y_test = test_df['text'], test_df['label']
print(f'Train size: {len(X_train)} | Test size: {len(X_test)}')
```

```python
models = {
    'Logistic Regression': LogisticRegression(C=1.5, random_state=42),
    'Multinomial Naive Bayes': MultinomialNB(alpha=0.1),
    'Linear SVM (Calibrated)': CalibratedClassifierCV(LinearSVC(dual='auto', random_state=42), cv=3),
    'SGD Classifier': SGDClassifier(loss='log_loss', random_state=42)
}

results = []
for name, model in models.items():
    pipe = Pipeline([('tfidf', TfidfVectorizer(ngram_range=(1, 2), stop_words='english')), ('clf', model)])
    pipe.fit(X_train, y_train)
    preds = pipe.predict(X_test)
    score = f1_score(y_test, preds)
    results.append({'Model': name, 'F1-Score': score})

pd.DataFrame(results)
```

### Conclusion:
All tested classifiers perform robustly on the curated WhatsApp corpus due to strong lexical indicators. Linear SVM and Logistic Regression offer top calibrated probability metrics suitable for explainability pipelines.
