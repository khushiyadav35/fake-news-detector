### Objective:
Demystify model predictions using Local Interpretable Model-agnostic Explanations (LIME).


```python
import sys
sys.path.append('..')
import joblib
from lime.lime_text import LimeTextExplainer
from src.predict import NewsPredictor
```

```python
predictor = NewsPredictor(model_path='../models/model_pipeline.pkl')
test_sample = 'URGENT: Drinking hot water with lemon kills coronavirus immediately! Share to all groups before midnight!'
result = predictor.predict(test_sample)
print('Predicted Label:', result['prediction'])
print('Confidence:', result['confidence'], '%')
```

```python
print('Features contributing to Suspicious:')
for word, weight in result['explanation']['suspicious_features']:
    print(f'  -> {word:15s} | Weight: {weight:.4f}')
```

### Why LIME Matters in Misinformation Moderation:
Black-box deep learning models lack interpretability. LIME perturbs input sentences and trains a local surrogate model, highlighting the exact trigger tokens so content moderators and users can verify the evidence directly.
